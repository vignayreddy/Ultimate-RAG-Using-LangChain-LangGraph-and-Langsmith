### Building Chatbot with Multiple Tolls Uisng Langgraph

AIM -- Create a chatbot with tool capabilities from arxiv,wikipedia search and some functions

```python
from langchain_community.tools import ArxivQueryRun,WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper,ArxivAPIWrapper
```

```python
api_wrapper_arxiv = ArxivAPIWrapper(top_k_results=2,doc_content_chars_max=500)
arxiv1 = ArxivQueryRun(api_wrapper=api_wrapper_arxiv)
```

```python
print(arxiv1.name)
```

```python
arxiv1.invoke(" Attention is all you need")
```

```python
api_wrapper_wiki = WikipediaAPIWrapper(top_k_results=2,doc_content_chars_max=500)
wiki = WikipediaQueryRun(api_wrapper=api_wrapper_wiki)
```

```python
wiki.invoke("Attention is all you need")
```

```python
from dotenv import load_dotenv
load_dotenv()
import os
os.environ["TAVILY_API_KEY"]=os.getenv("TAVILY_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain_community.tools.tavily_search import  TavilySearchResults
tavily =TavilySearchResults()
```

```python
tavily.invoke("Provide me current recent AI News")
```

```python
### Combine all the tools in the list
tools = [arxiv1,wiki,tavily]
```

```python
## Initialize my LLM Model
from langchain_groq import ChatGroq
llm = ChatGroq(model="llama-3.1-8b-instant")
```

```python
llm_with_tools = llm.bind_tools(tools)
from langchain_core.messages import HumanMessage,AIMessage
llm_with_tools.invoke([HumanMessage(content=f"What is the recent AI Tools")]).tool_calls
```
